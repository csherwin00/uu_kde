```python
import pandas as pd
import numpy as np
import seaborn as sns
df = pd.read_csv('data/heart_failure_clinical_records_dataset.csv')
df
```

```python
sns.kdeplot(df, x = 'ejection_fraction')

# help(sns.kdeplot)
```

```python
n = 15
samples = pd.DataFrame()
for i in range(n):
    sample = df.sample(frac = 1.0, replace = True)
    sns.kdeplot(sample, x = 'ejection_fraction')


```

# Q4 C

> The original KDE’s overall shape is reliable. The resampled curves show a main peak around 35–40 and a smaller peak around 60; however, the height of the main peak varies. The largest variation in density is 35–40. 

> There is also variation around 20–30 and 55–65. The curves are close at the tails, but this does not necessarily mean those regions are estimated correctly because there are few observations there.
